# EDA - fraud labels

Goals:
- Check label distribution and missing reasons
- Validate key uniqueness
- Verify coverage against orders


```python
import pandas as pd
from pathlib import Path

pd.set_option('display.max_columns', 200)
pd.set_option('display.width', 200)

ROOT = Path.cwd()
if not ((ROOT / 'fraud').exists() and (ROOT / 'e-commerce').exists()):
    if (ROOT.parent / 'fraud').exists() and (ROOT.parent / 'e-commerce').exists():
        ROOT = ROOT.parent

FRAUD_DIR = ROOT / 'fraud'
ECOM_DIR = ROOT / 'e-commerce'

fraud = pd.read_csv(FRAUD_DIR / 'fraud_labels.csv')
orders = pd.read_csv(ECOM_DIR / 'olist_orders_dataset.csv')
```

```python
print('fraud shape:', fraud.shape)
print('missing per column:')
print(fraud.isna().sum())
print('is_fraud distribution:')
print(fraud['is_fraud'].value_counts())
print('risk_score distribution:')
print(fraud['risk_score'].value_counts().sort_index().head(20))
print('label_source:')
print(fraud['label_source'].value_counts())
```

```python
reasons = fraud['fraud_reason'].fillna('')
reason_counts = reasons.str.split(';').explode().value_counts()
print('fraud_reason counts:')
print(reason_counts)
```

```python
orders_ids = set(orders['order_id'].unique())
fraud_ids = set(fraud['order_id'].unique())
print('fraud ids:', len(fraud_ids))
print('orders ids:', len(orders_ids))
print('fraud ids not in orders:', len(fraud_ids - orders_ids))
print('orders without fraud label:', len(orders_ids - fraud_ids))
```
